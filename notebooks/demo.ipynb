Summarize the synthetic job-search funnel.


```python
from pathlib import Path
    import pandas as pd

    data_path = Path("..") / "data" / "sample_synthetic_data.csv"
    df = pd.read_csv(data_path)
    display(df.head())
    display(df.describe(include="all").transpose().head(12))
    display(df.groupby("status")["application_id"].count().rename("count").sort_values(ascending=False))
```
